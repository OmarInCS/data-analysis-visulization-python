# Healthcare Data Collection Exercises

## Set 1: Exercises 1-5 (Interdependent)

### Exercise 1
Create a dictionary called `patients` where each key is a patient ID and the value is a tuple containing their name and room number. Add at least five patients to the dictionary.

- Example: `{'P001': ('John Doe', '101A'), 'P002': ('Jane Smith', '102B')}`

### Exercise 2
Add a new patient to the `patients` dictionary with the ID `'P006'`, name `'Alice Brown'`, and room number `'103C'`.

### Exercise 3
Create a named tuple called `VitalSigns` with fields `heart_rate`, `blood_pressure`, and `temperature`. Then add the `vital_signs` data for patient `'P001'` as a new key-value pair in the `patients` dictionary.

### Exercise 4
Access the room number and the `vital_signs` for patient `'P001'` and print a message that shows the patient's current room and their vital signs.

### Exercise 5
Update the `patients` dictionary to change patient `'P002'`'s room number to `'104D'`. Also, add the `vital_signs` data for `'P002'` with `heart_rate=80`, `blood_pressure='110/70'`, and `temperature=37.0`.




## Set 2: Exercises 6-10 (Interdependent)

### Exercise 6
Create a list called `medications` that contains dictionaries, where each dictionary stores the `patient_id`, `medication_name`, and `dosage`. Start with two entries for patients `'P003'` and `'P004'`.

- Example: `[{'patient_id': 'P003', 'medication_name': 'Aspirin', 'dosage': '100mg'}, ...]`

### Exercise 7
Add a new medication record to the `medications` list for patient `'P005'` who is prescribed `'Amoxicillin'` at a dosage of `'500mg'`.

### Exercise 8
Manually check and find the medications prescribed for patient `'P003'` from the `medications` list and print them.

### Exercise 9
Remove the medications for patient `'P003'`. If the patient is not found, print an appropriate message indicating that no records exist for the patient.

### Exercise 10
Create a new dictionary called `treatment_plan` that stores `patient_id` as the key and a list of medications as the value. Use the `medications` list to populate the `treatment_plan` dictionary for all patients.

# Cyber Security Data Collection Exercises

## Set 1: Exercises 1-5 (Interdependent)

### Exercise 1
Create a dictionary called `network_devices` where each key is a device ID and the value is a tuple containing the device type (e.g., 'Router', 'Firewall') and its IP address. Add at least five devices to the dictionary.

- Example: `{'D001': ('Router', '192.168.1.1'), 'D002': ('Firewall', '192.168.1.2')}`

### Exercise 2
Add a new network device to the `network_devices` dictionary with the ID `'D006'`, type `'Switch'`, and IP address `'192.168.1.3'`.

### Exercise 3
Create a named tuple called `DeviceStatus` with fields `uptime`, `cpu_usage`, and `memory_usage`. Then add the `device_status` data for device `'D001'` as a new key-value pair in the `network_devices` dictionary.

### Exercise 4
Access the IP address and the `device_status` for device `'D001'` and print a message showing the current status of the device, including uptime, CPU usage, and memory usage.

### Exercise 5
Update the `network_devices` dictionary to change the IP address of device `'D002'` to `'192.168.1.10'`. Also, add the `device_status` data for `'D002'` with `uptime='24 hours'`, `cpu_usage='30%'`, and `memory_usage='60%'`.


## Set 2: Exercises 6-8 (Interdependent)

### Exercise 6
Create a named tuple called `AccessLog` with fields `user_id`, `device_id`, and `timestamp`. Then, create an instance of `AccessLog` representing a login by user `'U100'` to device `'D003'` at timestamp `'2024-10-06 09:15:00'`.

### Exercise 7
Create a list called `access_logs` and add the `AccessLog` instance from Exercise 6 to it. Then, create two more `AccessLog` instances for different users and devices, and add them to the list.

### Exercise 8
