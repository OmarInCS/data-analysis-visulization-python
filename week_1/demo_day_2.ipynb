### 1. Reading Data with Pandas

#### 1. Reading CSV Files
The most straightforward way to read a CSV file is using the `pd.read_csv()` method.

**Key Parameters:**
* `filepath_or_buffer`: The path to the CSV file. It can be a local file path or a URL.
* `sep`: Specifies the delimiter. The default is a comma (`,`), but you can use other delimiters (e.g., `;`, `|`).
* `header`: Row number(s) to use as the column names. Default is 0 (first row).
* `names`: If column names are not present, you can pass a list of column names.
* `index_col`: Column(s) to set as the index of the DataFrame.

#### 2. Reading Excel Files
To read Excel files, use `pd.read_excel()`. This method supports multiple sheets, which is useful for more complex Excel files.

**Key Parameters:**
* `io`: Path to the Excel file.
* `sheet_name`: Specifies which sheet to read. Default is 0 (first sheet).
Can be an integer for sheet index, a string for sheet name, or a list of names/indices to load multiple sheets.
* `header`: Row number to use as the column names.
* `index_col`: Column(s) to set as the index of the DataFrame.
* `usecols`: Specify which columns to parse from the file.
